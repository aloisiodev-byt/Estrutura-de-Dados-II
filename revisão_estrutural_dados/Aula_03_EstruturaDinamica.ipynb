Exercício 2 — Construindo uma cadeia
Chorar:


Desafio - Sistema de Clínica;
Paciente: nome,

````python
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")
````

Conectar: A → B → C → None

Depois percorra a estrutura e produza:

````python
A
B
C
````

````python
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None


n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
````
